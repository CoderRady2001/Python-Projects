```python
num1=float(input("Enter Your First Number here:"))
num2=float(input("Enter Your second Number here"))

#The next step I will Potray the available options

print("\nSelect an operation:")
print("1.multiplication(*)")
print("2.addition(+)")
print("3.substraction(-)")
print("4.division(/)")

#Next I will ask the user which operation do they want to perfom

operation=input("Enter your choice of operation (1/2/3/4) or the operator (+/-/*/): ")

#After the user inputs his operation he/she wants to perfom then:

if operation in("1","*"):
    result= num1 * num2
    print(f"\ {num1} * {num2}={result}")
    print("Operation:Multiplication")
    
elif operation in ("2", "+"):
    result= num1 + num2
    print(f"\{num1}+ {num2}= {result}")
    print("Operation:Addition")
          
elif operation in("3","-"):
          result=num1-num2
          print(f"\{num1}-{num2}={result}")
          print("Operation:Substraction")

elif operation in("4","/"):
          if num2== 0:
              print(f"\Error: Cannot divide number by 0")
          else:
              result=num1/num2
              print(f"\{num1}/{num2}={result}")
              print(f"\operation:Division")

else:
    print(f"\Please Insert a valid Input")
          
```
