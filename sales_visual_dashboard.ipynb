```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv(r"C:\Users\Administrator\Downloads\sales_dataset(1).csv")

df
```

```python
df["Date"]=pd.to_datetime(df["Date"]) #This is converting the data column to date data type

df.isnull().sum() #I will check for missing values

df.dtypes #handling data types
```

```python
#SALES TRENDS OVER TIME(LINEPLOT)

plt.figure(figsize=(5,5))
sns.lineplot(x="Sales", y="Date", data=df)
plt.title("Sales Trends over Time")
plt.show()

print(" According to the visualization, the sales proved an increasing growth over the period of Time")
```

```python
#2 Regional sales Comparison(Bar Chart)

sns.barplot(x="Region",y="Sales", data=df, estimator= sum, palette="viridis"))
plt.title("Regional sales Comparison")
plt.show()

print("""The south Region leads with the Highest sales over all the regions followed by North Region.
The East Region has the least sales compared to all the other regions""")
```

```python
#Product Profitability

sns.boxplot(x="Product",y="Profit", data=df)

plt.title("Product Profitability")
plt.show()

print("""Printers are the most profitable category on average with the highest potential, while Laptops are the weakest. 
The company has a clear profitability hierarchy among these products, but high variation suggests room for process improvements to reduce low-profit outcomes.""")
```

```python
#CORRELATIONAL HEATMAP

corr=df.corr(numeric_only=True)

sns.heatmap(corr, annot=True, cmap= 'coolwarm', fmt=".2f", linewidths=0.5)
plt.title(" Sales vs Profit vs Quantity correlation")

plt.show()

print("Profit is entirely driven by Sales (perfect correlation of 1.00), not by Quantity (only moderate correlation of 0.50)")
```

```python
#Product-Region Faceted View

g = sns.FacetGrid(df, col="Region", hue="Product")
g.map(sns.scatterplot, "Sales", "Profit", alpha=0.7)
g.add_legend()
```
