```python
import pandas as pd
import numpy as np

#I will now load the dataset

df = pd.read_excel(r"C:\Users\Administrator\Downloads\Sales_data.xlsx")

#Step 2: I will now explore the data

df.head()
df.info()
df.describe()

#Next step I will check for either missing values, the datatypes and if there is any duplicates

print(df.isnull().sum())
print(df.dtypes)
print(df.duplicated().sum())

#STEP 3--- I will clean and prepare the sales Data

df.drop_duplicates(inplace=True)

#Next step I will analyze the KeyMetrics

df.groupby("Region")["Sales"].mean() # This will calculate the Region Average sales

#Next I will find the Highest and lowest Perfoming Producs

Highest_perfoming_Products=df.groupby("Product")["Sales"].sum().sort_values(ascending=False)

print("Highest perfoming product",Highest_perfoming_Products)


#Monthly sales Trends

df.groupby("Month")["Sales"].sum()

#Next step I will indentify the correlations

df.corr(numeric_only=True)

#Next I will summarize the Key Findings

print("=="*50)
print("KEY FINDINGS")

print("-Highest performing product: Laptop with 14,700 in sales – nearly 45% higher than the second-place product (Smartphone).")
print("-Sales concentration: The top 3 products (Laptop, Smartphone, Tablet) account for the majority of total sales.")
print("-Lowest performing products: Mouse (1,450) and Keyboard (1,850) – significantly behind other products.")

print("=="*50)
```
