```python
students= {}


#The next step I will create a function to calculate the grades

def letter_grade(avg):
    if avg>80:
        return 'A'
    elif avg >= 70:
        return 'B'
    elif avg >= 60:
        return 'C'
    elif avg >= 50:
        return 'D'
    elif avg >= 40:
        return 'E'
    else:
        return 'Fail'

#Input section

print("Welcome to the Student Grade Management System!")

num_students=int(input("enter number of the students: "))

for i in range(num_students):
    name= input(f"\n Enter Name of the student {i+1}:")
    
    grades = []
    
    geo= float(input("Enter marks for geo: "))
    kiswahili = float(input("Enter marks for kiswahili: "))
    computer_studies = float(input("Enter marks for Computer studies: "))
    Python=float(input("Input marks for Python: "))
    
    
    grades.extend([geo, kiswahili, computer_studies,Python])
    students[name] = grades
    
#I will now create the processing and output section

print("\n--- Student Report ---")
print(f"{'Name':<10}{'Average':<10}{'Grade'}")
print("-" * 30)

total_class_avg = 0

for name, grades in students.items():
    avg = sum(grades) / len(grades)
    grade = letter_grade(avg)
    total_class_avg += avg

    print(f"{name:<10}{avg:<10.1f}{grade}")

print("-" * 30)

class_avg = total_class_avg / len(students)
print(f"Class Average: {class_avg:.1f}%")

```
