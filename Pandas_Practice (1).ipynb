# PART 2 — Gaps to fill inYou've got a solid handle on exploration, `loc`/`iloc`, filtering, sorting, adding/dropping columns, missing data, descriptive stats, groupby/agg, pivot tables, string methods, and datetime.Below are things that weren't in your notebook yet. Since I don't have your actual `Global Superstore Data.xlsx`, these cells build a small sample dataframe with the same column names you used, so everything below runs on its own — but every method works exactly the same way on your real `df`.

```python
#Building a small sample dataset with the same columns you were using, so these cells run standalone

sample_data = {
    "Row ID": [1,2,3,4,5,6,7,8],
    "Order ID": ["US-001","US-001","US-002","US-003","US-003","US-004","US-005","US-005"],
    "Customer Name": ["Brian Lee","Brian Lee","Amy Chen","Mike Diaz","Mike Diaz","Sara Kim","Tom Reed","Tom Reed"],
    "Segment": ["Consumer","Consumer","Corporate","Home Office","Home Office","Consumer","Corporate","Corporate"],
    "Region": ["West","West","East","South","South","West","East","East"],
    "Category": ["Technology","Furniture","Technology","Office Supplies","Technology","Furniture","Technology","Office Supplies"],
    "Sales": [12500, 800, 15300, 250, 9800, 1200, 22000, 430],
    "Profit": [1800, -50, 2100, 60, 1500, 90, 3400, 40],
    "Order Date": pd.to_datetime(["2023-01-05","2023-01-05","2023-02-10","2023-02-20",
                                   "2023-02-20","2023-03-01","2023-03-15","2023-03-15"]),
    "Postal Code": [94105, 94105, 10001, 60601, 60601, 94105, 10001, None],
}

df = pd.DataFrame(sample_data)
df
```

## 1. Combining DataFrames — merge, join, concatThe biggest gap. Real projects almost always need to combine data from more than one table.

```python
#merge - SQL-style joins on a shared column (most common way to combine dataframes)

region_managers = pd.DataFrame({
    "Region": ["West","East","South","Central"],
    "Manager": ["Alicia","David","Priya","Noah"]
})

pd.merge(df, region_managers, on="Region", how="left")   #how: 'left','right','inner','outer'

#concat - stack dataframes on top of each other (or side by side)

df_jan = df[df["Order Date"].dt.month==1]
df_feb = df[df["Order Date"].dt.month==2]

pd.concat([df_jan, df_feb])              #stack rows (axis=0 by default)
pd.concat([df, region_managers], axis=1) #glue columns side by side (must align on index)

#join - like merge but joins on the INDEX by default
region_managers_idx = region_managers.set_index("Region")
df.set_index("Region").join(region_managers_idx, how="left")
```

## 2. value_counts — Quick Frequency Counts

```python
#value_counts - count how many times each unique value appears (extremely common in EDA)

df["Segment"].value_counts()
df["Segment"].value_counts(normalize=True)   #as proportions instead of raw counts
```

## 3. apply and map — Custom Row/Column Logic

```python
#apply - run a custom function across a column, row, or the whole dataframe

df["Sales"].apply(lambda x: x*1.1)                  #applied to a single column
df.apply(lambda row: row["Sales"]-row["Profit"], axis=1)  #applied across rows (axis=1)

#map - element-wise substitution/transformation, usually for a single Series

segment_short = {"Consumer":"CON","Corporate":"CORP","Home Office":"HO"}
df["Segment"].map(segment_short)
```

## 4. isin and between — Cleaner Filtering

```python
#isin - filter rows where a column matches any value in a list (cleaner than chaining | )

df[df["Region"].isin(["West","East"])]

#between - filter numeric ranges

df[df["Sales"].between(1000, 20000)]
```

## 5. set_index and reset_index

```python
#set_index - make a column the row index (useful for lookups, joins, time series)

df_indexed = df.set_index("Order ID")
df_indexed

#reset_index - turn the index back into a regular column

df_indexed.reset_index()
```

## 6. astype — Converting Data Types

```python
#astype - explicitly convert a column's dtype

df["Row ID"].astype(str)
df["Sales"].astype(float)
df["Segment"].astype("category")   #category dtype saves memory for repeated text values
```

## 7. duplicated and replace

```python
#duplicated - flags duplicate rows (you already used drop_duplicates, this is the check before dropping)

df.duplicated(subset="Order ID")          #True where a row repeats an earlier Order ID
df[df.duplicated(subset="Order ID", keep=False)]  #show ALL rows involved in a duplicate

#replace - swap specific values anywhere in a column (more flexible than str.replace)

df["Category"].replace({"Office Supplies":"Supplies"})
```

## 8. Binning Numbers — cut and qcut

```python
#cut - bin numeric values into custom ranges (great for turning Sales into categories)

df["Sales_Tier"] = pd.cut(df["Sales"], bins=[0,1000,10000,25000], labels=["Low","Medium","High"])
df[["Sales","Sales_Tier"]]

#qcut - bin into equal-sized groups (quartiles, etc.) instead of custom ranges

pd.qcut(df["Sales"], q=2, labels=["Bottom Half","Top Half"])
```

## 9. nlargest and nsmallest

```python
#nlargest/nsmallest - quick top-N / bottom-N rows, faster and cleaner than sort_values().head()

df.nlargest(3, "Sales")
df.nsmallest(3, "Sales")
```

## 10. Grouping by Multiple Columns + transform

```python
#Grouping by more than one column

df.groupby(["Region","Segment"])["Sales"].sum()

#transform - like groupby+agg, but returns a result the SAME SIZE as the original dataframe
#useful for things like "each row's sales as a % of its region's total"

df["Region_Total_Sales"] = df.groupby("Region")["Sales"].transform("sum")
df["Pct_of_Region"] = df["Sales"] / df["Region_Total_Sales"]
df[["Region","Sales","Region_Total_Sales","Pct_of_Region"]]
```

## 11. crosstab — Quick Cross-Tabulation

```python
#crosstab - frequency table between two columns, similar to pivot_table but simpler for counts

pd.crosstab(df["Region"], df["Category"])
pd.crosstab(df["Region"], df["Category"], values=df["Sales"], aggfunc="sum")
```
