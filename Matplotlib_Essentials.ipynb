```python
#DATA VISUALIZATION--- This is one of the most powerful tools in data science, it helps to transform raw data into clear and meaningful visuals that help communicate insights clearly

import matplotlib.pyplot as plt
import seaborn as sns
sns.set_palette("colorblind")

months=['Jan','Feb','Mar','Apr','May','Jun']
sales=[12000,15000,17000,13000,19000,21000]

plt.plot(months,sales, marker ='o',color='teal')
plt.title("Monthly Sales in 2024")
plt.xlabel("Month")
plt.ylabel("Sales($)")

plt.show()
```

```python
#LINE PLOTS--- A line plot is ideal for showing trends over time or continous relationships

import matplotlib.pyplot as plt

months=["Jan","Feb","Mar","April","May"]
sales=[1000,1200,900,1400,1700]

plt.plot(months, sales,color='teal',marker='o',linestyle='--')
plt.title("Monthly sales")
plt.xlabel("Month")
plt.ylabel("Sales ($)")
plt.grid(True)
plt.show()
```

```python
#HISTOGRAMS-- Shows the distribution of numerical data which is useful for exploring ranges and frequencies

import numpy as np
import matplotlib.pyplot as plt
data=np.random.randn(1000) # 1000 random values

plt.hist(data,bins=20,color="mediumseagreen",edgecolor="black")
plt.title("Data Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()

#A bell shaped curve suggests a normal distribution, common in real-world datasets
```

```python
#SUBPLOTS AND MULTIPLE PLOTS--one can create charts in a single figure for easy comparison

import matplotlib.pyplot as plt
import numpy as np

x=np.linspace(0,10,100)

fig,axes=plt.subplots(1,2,figsize=(10,4))

axes[0].plot(x,np.sin(x),color='teal')
axes[0].set_title("sine Wave")

axes[1].plot(x,np.cos(x),color='orange')
axes[1].set_title("cosine Wave")

plt.tight_layout()  #This helps preventing the labels from overlapping
plt.show()
```

```python
#BOX PLOTS AND VIOLIN PLOTS--when working with statistical or experimental data, it is important to understand distribution, spread and outliers
#A box plot shows median(median value),quartiles(q1 & q3),outliers(points beyond whiskers)

import matplotlib.pyplot as plt
import numpy as np

data=[np.random.normal(50,10,200),np.random.normal(60,15,200),np.random.normal(70,5,200)]

plt.boxplot(data,patch_artist=True,boxprops=dict(facecolor='lightblue'))
plt.title("Box plot of sample distribution")
plt.xlabel("Groups")
plt.ylabel("Values")
plt.show()

#If one box is much taller than another, the the dataset had more variability
#Dots outside the whiskers represents outliers
```

```python
#VIOLIN PLOT

plt.violinplot(data,showmeans=True)
plt.title("Violin Plot of Distribution")
plt.xlabel("Groups")
plt.ylabel("Values")
plt.show()

#Use violin plots when you want to compare data symmetry or multi-modal distribution(multiple peaks)
```

```python
#Heatmaps--They are perfect for visualizing correlations or intensity values in 2D form-often used in machine Learning or feature analysis

import seaborn as sns
import numpy as np
#simulated correlation data

data=np.random.rand(5,5)
plt.figure(figsize=(6,5))
sns.heatmap(data,annot=True, cmap="coolwarm",fmt=".2f")
plt.title("Sample Heatmap")
plt.show()
```

```python
#Logarithmic Sclaes-- when your data spans large numeric ranges(e.g population growth,GDP, income) linear scales hide patterns
#Logarithmic scales make trends clearer

import numpy as np

x=np.linspace(1,1000,100)
y=x**2

plt.plot(x,y)
plt.xscale('log')
plt.yscale('log')
plt.title('Logarithmic Scale Plot')
plt.xlabel("Log X")
plt.ylabel("log Y")

plt.show()

#Log scales helps to reveal proportional growth and exponential growth
```

```python
#3D plots--matplotlibs also support 3D visalizations through its axes3D module-useful for visualizing multideimensional data

from mpl_toolkits.mplot3d import Axes3D
import matplotlib.pyplot as plt
import numpy as np

fig=plt.figure()
ax=fig.add_subplot(111,projection='3d')

x=np.linspace(-5,5,50)
y=np.linspace(-5,5,50)

x,y=np.meshgrid(x,y)
z=np.sin(np.sqrt(x**2 + y**2))

ax.plot_surface(x,y,z,cmap='viridis')
ax.set_title("3D surface plot example")

plt.show()
```

## 1. Annotations and TextLabeling specific points on a chart — very common when highlighting a peak, an outlier, or a key event.

```python
import matplotlib.pyplot as plt

months=['Jan','Feb','Mar','Apr','May','Jun']
sales=[12000,15000,17000,13000,19000,21000]

plt.plot(months,sales,marker='o',color='teal')
plt.title("Monthly Sales in 2024")

#annotate - point an arrow at a specific data point with a label
plt.annotate('Peak sales', xy=('Jun',21000), xytext=('Apr',20500),
             arrowprops=dict(facecolor='black', arrowstyle='->'))

#text - just place text anywhere on the chart, no arrow
plt.text('Jan',12500,'Start of year', fontsize=9, color='gray')

plt.show()
```

## 2. Error BarsShowing uncertainty or variability around a measurement — common in scientific/experimental plots.

```python
import numpy as np

x=np.arange(5)
y=[20,25,22,28,24]
error=[2,1.5,3,2,1]

plt.errorbar(x,y,yerr=error,fmt='o-',color='teal',ecolor='gray',capsize=5)
plt.title("Measurements with Error Bars")
plt.xlabel("Trial")
plt.ylabel("Value")
plt.show()
```

## 3. fill_between — Area/Shaded Region Plots

```python
x=np.linspace(0,10,100)
y=np.sin(x)

plt.plot(x,y,color='teal')
plt.fill_between(x,y,0,alpha=0.3,color='teal')   #shades the area under the curve
plt.title("Area Under a Curve")
plt.show()

#Useful for confidence bands too - shading between a lower and upper bound
y_upper=y+0.2
y_lower=y-0.2

plt.plot(x,y,color='darkorange')
plt.fill_between(x,y_lower,y_upper,alpha=0.2,color='darkorange')
plt.title("Line with Confidence Band")
plt.show()
```

## 4. Grouped and Stacked Bar ChartsYou covered a single bar chart, but comparing multiple series side by side or stacked is very common.

```python
products=["Apples","Bananas","Mango","Oranges"]
sales_2023=[300,450,150,400]
sales_2024=[350,420,200,380]

x=np.arange(len(products))
width=0.35

#Grouped bars - two bars side by side per category
fig,ax=plt.subplots()
ax.bar(x-width/2,sales_2023,width,label='2023',color='skyblue')
ax.bar(x+width/2,sales_2024,width,label='2024',color='coral')
ax.set_xticks(x)
ax.set_xticklabels(products)
ax.set_title("Sales Comparison by Year")
ax.legend()
plt.show()

#Stacked bars - one bar per category, segments stacked on top of each other
fig,ax=plt.subplots()
ax.bar(products,sales_2023,label='2023',color='skyblue')
ax.bar(products,sales_2024,bottom=sales_2023,label='2024',color='coral')
ax.set_title("Total Sales, Stacked by Year")
ax.legend()
plt.show()
```

## 5. Twin Axes — Two Y-Axes on One ChartUseful when comparing two variables with very different scales (e.g. temperature vs. rainfall).

```python
months=['Jan','Feb','Mar','Apr','May']
temperature=[15,17,20,23,27]
rainfall=[80,65,50,40,30]

fig,ax1=plt.subplots()

ax1.plot(months,temperature,color='darkorange',marker='o')
ax1.set_xlabel("Month")
ax1.set_ylabel("Temperature (C)",color='darkorange')

ax2=ax1.twinx()   #creates a second y-axis sharing the same x-axis
ax2.bar(months,rainfall,alpha=0.3,color='steelblue')
ax2.set_ylabel("Rainfall (mm)",color='steelblue')

plt.title("Temperature vs Rainfall")
plt.show()
```

## 6. Custom Axis Limits and Ticks

```python
x=np.linspace(0,10,100)
y=np.sin(x)

plt.plot(x,y,color='teal')
plt.xlim(0,8)                       #zoom in on the x-axis range
plt.ylim(-1.5,1.5)                  #control the y-axis range
plt.xticks([0,2,4,6,8],['Start','Q1','Q2','Q3','End'])  #custom tick labels
plt.yticks([-1,0,1])
plt.title("Custom Axis Limits and Tick Labels")
plt.show()
```

## 7. Scatter Plots with a Color Map (Bubble Charts)You covered a basic scatter plot — here's how to encode a THIRD variable using color and a FOURTH using size.

```python
np.random.seed(1)
x=np.random.rand(50)*100
y=np.random.rand(50)*100
sizes=np.random.rand(50)*300      #a 4th variable via marker size
colors=np.random.rand(50)         #a 3rd variable via color

plt.scatter(x,y,c=colors,s=sizes,cmap='viridis',alpha=0.7)
plt.colorbar(label='Intensity')   #shows what the colors mean
plt.title("Bubble Chart: Position, Color, and Size Encode 4 Variables")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

## 8. Stem and Step PlotsAlternatives to line plots, useful for discrete signals or cumulative/binned data.

```python
x=np.arange(10)
y=np.random.randint(1,10,10)

plt.stem(x,y)
plt.title("Stem Plot")
plt.show()

plt.step(x,y,where='mid',color='teal')
plt.title("Step Plot")
plt.show()
```

## 9. Contour PlotsVisualizing a 3D surface as 2D contour lines — the 2D cousin of the 3D surface plot you already did.

```python
x=np.linspace(-5,5,100)
y=np.linspace(-5,5,100)
X,Y=np.meshgrid(x,y)
Z=np.sin(np.sqrt(X**2+Y**2))

plt.contourf(X,Y,Z,cmap='viridis')   #filled contour
plt.colorbar(label='Z value')
plt.title("Contour Plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

## 11. Dates on the X-Axis

```python
import matplotlib.dates as mdates
import datetime

dates=[datetime.date(2024,1,1)+datetime.timedelta(days=30*i) for i in range(6)]
values=[100,120,115,140,160,155]

fig,ax=plt.subplots()
ax.plot(dates,values,marker='o',color='teal')

ax.xaxis.set_major_formatter(mdates.DateFormatter('%b %Y'))  #format as 'Jan 2024'
fig.autofmt_xdate()   #auto-rotates date labels so they don't overlap

plt.title("Values Over Time")
plt.show()
```

## 12. Complex Layouts with GridSpecYou covered plt.subplots() for even grids — GridSpec lets panels span different rows/columns for dashboard-style layouts.

```python
fig=plt.figure(figsize=(8,6))
gs=fig.add_gridspec(2,2)

ax1=fig.add_subplot(gs[0,:])    #top row, spans both columns
ax2=fig.add_subplot(gs[1,0])    #bottom-left
ax3=fig.add_subplot(gs[1,1])    #bottom-right

ax1.plot(np.linspace(0,10,100),np.sin(np.linspace(0,10,100)),color='teal')
ax1.set_title("Wide Top Panel")

ax2.bar(['A','B','C'],[3,7,5],color='coral')
ax2.set_title("Bottom Left")

ax3.scatter(np.random.rand(20),np.random.rand(20),color='purple')
ax3.set_title("Bottom Right")

plt.tight_layout()
plt.show()
```

## 13. Global Style Settings with rcParamsAn alternative to plt.style.use() when you want fine-grained control over defaults for every plot in the notebook.

```python
plt.rcParams['figure.figsize']=(6,4)
plt.rcParams['font.size']=11
plt.rcParams['axes.grid']=True
plt.rcParams['lines.linewidth']=2

#every plot from here on will use these defaults automatically
plt.plot(np.linspace(0,10,50),np.cos(np.linspace(0,10,50)),color='darkorange')
plt.title("Plot Using Custom rcParams Defaults")
plt.show()

#reset back to matplotlib defaults if needed
plt.rcParams.update(plt.rcParamsDefault)
```
