## 1. Universal Functions (ufuncs)Fast, element-wise math functions. These are the building blocks behind most vectorized operations you already used (like `prices*(1+tax_rate)`).

```python
#Universal Functions (ufuncs) - element wise math applied without loops

arr = np.array([1, 4, 9, 16])

np.sqrt(arr)          #square root of each element
np.exp(arr)           #e^x for each element
np.log(arr)           #natural log
np.log10(arr)         #base-10 log

angles = np.array([0, np.pi/2, np.pi])
np.sin(angles)        #sine of each angle (radians)
np.cos(angles)

np.round(np.array([1.234, 5.678]), 2)  #round to 2 decimals
np.abs(np.array([-3, -1, 2]))          #absolute value
```

## 2. Sorting and Searching

```python
#Sorting and Searching

scores = np.array([55, 92, 40, 78, 65])

np.sort(scores)          #returns a new sorted array (ascending)
np.argsort(scores)       #indices that WOULD sort the array - very useful for ranking

#e.g. get names sorted by score
students = np.array(['A','B','C','D','E'])
order = np.argsort(scores)
students[order]

#searchsorted - find where a value would be inserted to keep the array sorted
sorted_scores = np.sort(scores)
np.searchsorted(sorted_scores, 70)
```

## 3. np.unique and Set Operations

```python
#np.unique - get distinct values (and optionally their counts)

data = np.array([1, 2, 2, 3, 3, 3, 4])

np.unique(data)                              #[1,2,3,4]
values, counts = np.unique(data, return_counts=True)
dict(zip(values, counts))                    #how many times each value appears

#Set operations between two arrays

a = np.array([1,2,3,4])
b = np.array([3,4,5,6])

np.intersect1d(a,b)   #values in both
np.union1d(a,b)       #all unique values from both
np.setdiff1d(a,b)     #in a but not in b
```

## 4. Cumulative Operations and Differences

```python
#Cumulative sum / product, and differences between consecutive elements

sales = np.array([100, 150, 200, 130])

np.cumsum(sales)   #running total: [100,250,450,580]
np.cumprod(np.array([1,2,3,4]))  #running product: [1,2,6,24]

np.diff(sales)      #difference between consecutive elements: [50,50,-70]
                     #handy for day-over-day change, velocity from position, etc.
```

## 5. Adding/Removing Dimensions

```python
#Sometimes you need to add or remove a dimension to make shapes compatible (e.g. for broadcasting)

a = np.array([1,2,3])       #shape (3,)

a[:, np.newaxis]            #shape (3,1) - turns a row vector into a column vector
np.expand_dims(a, axis=0)   #shape (1,3) - same idea, more explicit function form

b = np.array([[[1,2,3]]])   #shape (1,1,3)
np.squeeze(b)                #removes all size-1 dimensions -> shape (3,)
```

## 7. Comparing Arrays Safely

```python
#Comparing arrays - watch out for floating point precision issues

a = np.array([1.0, 2.0, 3.0])
b = np.array([1.0, 2.0, 3.0000001])

a == b                 #element-wise comparison, unreliable for floats
np.array_equal(a,b)    #True only if ALL elements match exactly
np.allclose(a,b)       #True if all elements are close within a tolerance - the safer choice for floats
np.isclose(a,b)        #element-wise version of allclose
```

## 10. repeat vs tile

```python
#repeat vs tile - both duplicate elements, but differently

a = np.array([1,2,3])

np.repeat(a, 2)   #repeats EACH element in place -> [1,1,2,2,3,3]
np.tile(a, 2)     #repeats the WHOLE array -> [1,2,3,1,2,3]
```

## 12. Why Vectorize? Loops vs NumPy, Timed

```python
#The whole point of NumPy: vectorized operations are much faster than plain Python loops

import time

n = 1_000_000
python_list = list(range(n))
numpy_array = np.arange(n)

start = time.time()
squared_list = [x**2 for x in python_list]
print("Python loop:", time.time() - start, "seconds")

start = time.time()
squared_array = numpy_array ** 2
print("NumPy vectorized:", time.time() - start, "seconds")

#NumPy is typically 10-100x faster because it runs in compiled C code under the hood,
#instead of interpreting each element one at a time
```
